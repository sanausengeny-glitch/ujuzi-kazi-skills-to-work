# Ujuzi Kazi – Skills to Work

## Project Overview

Ujuzi Kazi is a youth employment and skills-matching platform designed to connect young people with relevant job opportunities based on their education, technical skills, experience, and location.

## Project Objectives

- Identify skills gaps among young people.
- Analyse available job opportunities.
- Understand youth employment and unemployment patterns.
- Match youth skills with employer requirements.
- Provide insights that can support better employment opportunities.

## Data Sources

This project uses two sample datasets:

1. `job_adverts.csv` – contains job opportunities and employer skill requirements.
2. `youth_profiles.csv` – contains youth education, skills, experience, and employment information.

**Data note:** The datasets are synthetic sample data created for demonstration and analysis purposes. They are not verified records of actual job advertisements or individual young people.

## Tools Used

- Python
- Pandas
- Matplotlib
- Google Colab
- GitHub

```python
# Import the required libraries

import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Load the project datasets from GitHub

jobs_url = "https://raw.githubusercontent.com/sanausengeny-glitch/ujuzi-kazi-skills-to-work/main/data/job_adverts.csv"

youth_url = "https://raw.githubusercontent.com/sanausengeny-glitch/ujuzi-kazi-skills-to-work/main/data/youth_profiles.csv"

jobs = pd.read_csv(jobs_url)
youth = pd.read_csv(youth_url)

print("Job Advertisements Dataset:")
display(jobs.head())

print("Youth Profiles Dataset:")
display(youth.head())
```

```python
# Check dataset size and structure

print("JOB ADVERTISEMENTS")
print("Shape:", jobs.shape)
print("\nColumn information:")
jobs.info()

print("\nYOUTH PROFILES")
print("Shape:", youth.shape)
print("\nColumn information:")
youth.info()
```

```python
# Analyse youth employment status

employment_counts = youth["employment_status"].value_counts()

print("Youth Employment Status:")
display(employment_counts)

# Calculate percentages
employment_percentages = (
    youth["employment_status"].value_counts(normalize=True) * 100
)

print("\nEmployment Status Percentages:")
display(employment_percentages.round(1))
```

```python
# Visualise youth employment status

plt.figure(figsize=(8, 5))

employment_counts.plot(
    kind="bar",
    color=["skyblue", "orange", "green"]
)

plt.title("Youth Employment Status")
plt.xlabel("Employment Status")
plt.ylabel("Number of Young People")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Analyse the technical skills of young people

skill_counts = youth["technical_skill"].value_counts()

print("Youth Technical Skills:")
display(skill_counts)

# Visualise technical skills

plt.figure(figsize=(10, 6))

skill_counts.plot(kind="bar")

plt.title("Technical Skills Among Young People")
plt.xlabel("Technical Skills")
plt.ylabel("Number of Young People")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Analyse employer skill requirements

demanded_skills = jobs["required_skill"].value_counts()

print("Skills Required by Employers:")
display(demanded_skills)

# Visualise employer skill demand

plt.figure(figsize=(10, 6))

demanded_skills.plot(kind="bar")

plt.title("Skills Demanded by Employers")
plt.xlabel("Required Skills")
plt.ylabel("Number of Job Advertisements")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Compare youth skills with employer requirements

youth_skills = set(youth["technical_skill"].str.lower())
job_skills = set(jobs["required_skill"].str.lower())

matching_skills = youth_skills.intersection(job_skills)

print("Skills available among youth:", len(youth_skills))
print("Skills required by employers:", len(job_skills))

print("\nDirectly Matching Skills:")
for skill in sorted(matching_skills):
    print("-", skill)

print("\nSkills that may need further training:")
for skill in sorted(job_skills - youth_skills):
    print("-", skill)
```

```python
# Analyse job opportunities by employment type

employment_types = jobs["employment_type"].value_counts()

print("Available Job Types:")
display(employment_types)

# Visualise job types

plt.figure(figsize=(8, 5))

employment_types.plot(kind="bar")

plt.title("Job Opportunities by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Analyse job opportunities by location

location_counts = jobs["location"].value_counts()

print("Job Opportunities by Location:")
display(location_counts)

# Create a location chart

plt.figure(figsize=(8, 5))

location_counts.plot(kind="bar")

plt.title("Job Opportunities by Location")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Analyse education requirements

education_counts = jobs["education_level"].value_counts()

print("Education Requirements:")
display(education_counts)

# Visualise education requirements

plt.figure(figsize=(8, 5))

education_counts.plot(kind="bar")

plt.title("Education Levels Required by Employers")
plt.xlabel("Education Level")
plt.ylabel("Number of Job Advertisements")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Analyse job opportunities by industry

industry_counts = jobs["industry"].value_counts()

print("Job Opportunities by Industry:")
display(industry_counts)

# Visualise industry distribution

plt.figure(figsize=(9, 5))

industry_counts.plot(kind="bar")

plt.title("Job Opportunities by Industry")
plt.xlabel("Industry")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Key Findings and Insights

Based on the analysis of the 10 sample youth profiles and 10 sample job advertisements, the following findings were identified:

### 1. Youth Employment
- 60% of the youth sample is unemployed.
- 20% is employed.
- 20% is self-employed.

### 2. Skills Gap
There are differences between the technical skills represented among young people and the skills requested by employers. Some exact matches include Python, Computer Skills and Data Annotation.

### 3. Employment Opportunities
The sample job advertisements include full-time, internship, freelance and contract opportunities.

### 4. Geographic Distribution
The job advertisements cover Nairobi, Narok, Kisumu and remote opportunities.

### 5. Industry Demand
The sample includes opportunities in ICT, marketing, creative services, supply chain, technology and research.

### Recommendations

1. Provide practical skills training based on employer requirements.
2. Help young people identify suitable job opportunities.
3. Encourage employers to offer internships and entry-level opportunities.
4. Improve access to remote and freelance opportunities.
5. Strengthen connections between training institutions and employers.
